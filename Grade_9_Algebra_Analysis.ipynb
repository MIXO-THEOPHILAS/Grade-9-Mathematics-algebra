```python
import pandas as pd

file_name = "GRADE 9 DATASET-simplyifying algebraic expression.ods"

df = pd.read_excel(
    file_name,
    engine="odf",
    header=1
)

df.head()
```

```python
print("Number of learners:", len(df))
print("Number of questions:", 15)
```

```python
print("Highest mark:", df["TOTAL MARKS"].max())
print("Lowest mark:", df["TOTAL MARKS"].min())
print("Average mark:", df["TOTAL MARKS"].mean())
```

```python
df[["LNAME", "TOTAL MARKS"]]
```

```python
df["PERCENTAGE"] = (df["TOTAL MARKS"] / 30) * 100

df[["LNAME", "TOTAL MARKS", "PERCENTAGE"]]
```

```python
df = df[df["LNAME"].astype(str).str.strip() != "TOTAL"].copy()

print(df.tail())
print("Number of learners:", len(df))
```

```python
question_cols = [f"Q{i}" for i in range(1, 16)]

print(question_cols)
```

```python
results = []

for question in question_cols:
    results.append({
        "Question": question,
        "Full marks (2)": (df[question] == 2).sum(),
        "Partial marks (1)": (df[question] == 1).sum(),
        "Zero marks (0)": (df[question] == 0).sum(),
        "Blank (B)": (df[question].astype(str).str.upper() == "B").sum()
    })

question_analysis = pd.DataFrame(results)

question_analysis
```

```python
percentage_analysis = question_analysis.copy()

categories = [
    "Full marks (2)",
    "Partial marks (1)",
    "Zero marks (0)",
    "Blank (B)"
]

for category in categories:
    percentage_analysis[category] = (
        percentage_analysis[category] / len(df) * 100
    ).round(2)

percentage_analysis
```

```python
numeric_questions = df[question_cols].apply(
    pd.to_numeric,
    errors="coerce"
)

question_means = numeric_questions.mean().round(2)

question_means
```

```python
# Define the mark ranges
bins = [0, 9, 14, 19, 24, 30]
labels = ["0–9", "10–14", "15–19", "20–24", "25–30"]

# Assign each learner to a mark range
df["MARK RANGE"] = pd.cut(
    df["TOTAL MARKS"],
    bins=bins,
    labels=labels,
    include_lowest=True
)

# Count learners in each range
mark_distribution = df["MARK RANGE"].value_counts().sort_index()

# Display the table
print(mark_distribution)
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import os

# Mark distribution from the results
mark_ranges = ["0–9", "10–14", "15–19", "20–24", "25–30"]
number_of_learners = [4, 3, 2, 15, 8]

# Create the table
distribution_table = pd.DataFrame({
    "Mark Range": mark_ranges,
    "Number of Learners": number_of_learners
})

# Calculate percentages
distribution_table["Percentage"] = (
    distribution_table["Number of Learners"] / 32 * 100
).round(2)

# Display the table
display(distribution_table)

# Create the figures folder if it does not exist
os.makedirs("figures", exist_ok=True)

# Create the graph
plt.figure(figsize=(9, 5))

plt.bar(
    distribution_table["Mark Range"],
    distribution_table["Number of Learners"]
)

plt.xlabel("Total Mark Range")
plt.ylabel("Number of Learners")
plt.title("Distribution of Learners by Total Test Mark")
plt.grid(axis="y", alpha=0.3)

plt.tight_layout()

# Save the graph
plt.savefig(
    "figures/mark_distribution.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

print("Graph saved successfully as: figures/mark_distribution.png")
```
